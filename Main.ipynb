# Interpretation

The dataset covers the Top 15 happiest countries in the World Happiness Report 2023.

**Ladder Score (Happiness Score):**
- Ranges from **6.894** (United States) to **7.804** (Finland)
- Mean score is **7.27** with a low standard deviation of **0.26**, indicating that the top 15 countries are tightly clustered in terms of happiness
- The 25th–75th percentile range (7.10 – 7.44) confirms these nations share a consistently high level of citizen wellbeing

**Healthy Life Expectancy:**
- Ranges from **65.85 years** (United States) to **72.90 years** (Switzerland)
- Mean is **71.20 years**, with a standard deviation of **1.62 years**
- The narrow spread suggests most top-ranked countries provide similarly high life expectancy, with the United States being a notable exception at the lower end

> ## To fulfill my Objective, I will be using many statistical graphs and charts

```python
# Replace blank spaces with NaN
df = df.replace(" ", np.nan)

# Fill NaN values in numeric columns with column means
df = df.fillna(df.mean(numeric_only=True))

# Verify
print(df.isnull().sum().sum())
```

## Graphs I will be making are :
1. Histogram
2. KDE
3. Scatterplot
4. Lineplot
5. Boxplot
6. Heatmap

# 1. Histogram

```python
sns.displot(df["Ladder score"])
```

# Interpretation
>  For ladder score,
> - Ladder scores range from 6.9 to 7.8
> - Most countries have a score between 7.0 and 7.3
> - The distribution is roughly bell-shaped (normal)

```python
sns.displot(df["Healthy life expectancy"])
```

# Interpretation
> For Healthy Life Expectancy,
> - Healthy Life expectancy ranges from 65.8 to 72.9 years
> - Most countries are clustered between 71 and 73 years
>- The United States (65.8 years) stands far below the rest

# 2. KDE

```python
sns.kdeplot(df["Ladder score"])
```

# Interpretation

> - The curve peaks around 7.1–7.3, where most countries lie
> - Slightly right-tailed due to Finland's high score of 7.8
> - Overall close to a normal distribution


```python
sns.kdeplot(df["Healthy life expectancy"])
```

# Interpretation
> - The curve peaks around 71–72 years
> - A long left tail exists because of the US outlier (65.8 years)
> - This makes the distribution left-skewed

# 3. Scatterplot

```python
sns.scatterplot(x=df['Healthy life expectancy'],y=df['Ladder score'])
```

# Interpretation

- Countries with higher Healthy Life expectancy tend to have higher happiness scores
- The US is the clear bottom-left outlier (low Healthy Life Expectancy, low score)
- The trend is positive but not perfectly straight

# 4. Lineplot

```python
sns.lineplot(x='Healthy life expectancy',y='Ladder score', data=df )
```

# Interpretation

> - As Healthy life expectancy increases, Ladder Score generally rises too
> - The line is not perfectly smooth, showing that other factors also affect happiness
> - There is a big jump from the US to the rest of the countries

# 5. Boxplot

```python
sns.boxplot(data=df, x="Healthy life expectancy", y="Ladder score")
```

# Interpretation

> For ladder Score,
> - Median happiness score is 7.24
> - Most countries fall between 7.1 and 7.4
> - No outliers, all 15 countries are relatively close to each other

> For Healthy Life Expectancy,
> - Median Healthy life expectancy score is 71.4 years
> - Most countries are tightly packed between 71.1 and 71.9 years
> - The United States (65.8) is a clear outlier, far below the group

# 6. Heatmap

```python
corrmat = df.select_dtypes(include=np.number).corr()

f, ax = plt.subplots(figsize=(16, 12))
sns.heatmap(corrmat, vmax=.8, square=True, annot=True)
plt.show()
```

# Interpretation

> - Social Support (0.63) and Freedom (0.60) are most strongly linked to happiness
> - Healthy Life Expectancy has a moderate positive link with happiness (0.45)
> - Corruption perception is negatively linked to less corruption = more happiness

# CONCLUSION:

The objective of this project was to find the relationship between "Ladderscore" and "Healthy Life Expectancy" for the top 15 countries in the world as of 2023.

Based on the EDA performed:

- The scatterplot and line plot both show a positive trend: countries with higher Healthy Life Expectancy tend to have higher happiness scores.

- The correlation value of 0.45 confirms a moderate positive relationship between the two variables.

- The United States was a consistent outlier throughout the analysis,
  having both the lowest Healthy Life expectancy (65.8 years) and the lowest Ladder Score (6.894) in this group.

Therefore, the objective was fulfilled. There is a moderate positive relationship between Healthy Life Expectancy and Ladder Score. However, Healthy Life Expectancy alone does not fully determine happiness; social support and Freedom also play a significant role, as seen in the heatmap.
